```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
delivery = pd.read_csv("../data/processed/delivery_clean.csv")
```

```python
#best and worst time of day
tod = delivery.groupby("time_of_day").agg(
    avg_delivery=("delivery_time_min","mean"),
    avg_prep=("preparation_time_min","mean"),
    orders=("delivery_time_min","count")
).reset_index().sort_values("avg_delivery", ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(data=tod, x="time_of_day", y="avg_delivery", palette="coolwarm", ax=ax)
ax.set_title("Average Delivery Time by Time of Day")
ax.set_ylabel("Avg Delivery Time (min)")
ax.set_xlabel("")
plt.tight_layout()
plt.savefig("../output/charts/06_delivery_by_time.png", dpi=150)
plt.show()
```

```python
#Traffic impact
traffic = delivery.groupby("traffic_level")["delivery_time_min"].mean().reset_index()

fig, ax = plt.subplots(figsize=(7, 5))
sns.barplot(data=traffic, x="traffic_level", y="delivery_time_min",
            palette="Reds_r", ax=ax)
ax.set_title("Delivery Time by Traffic Level")
ax.set_ylabel("Avg Delivery Time (min)")
plt.tight_layout()
plt.savefig("../output/charts/07_delivery_by_traffic.png", dpi=150)
plt.show()
```

```python
#Weather impact
weather = delivery.groupby("weather")["delivery_time_min"].mean().reset_index()

fig, ax = plt.subplots(figsize=(7, 5))
sns.barplot(data=weather, x="weather", y="delivery_time_min",
            palette="Blues_r", ax=ax)
ax.set_title("Delivery Time by Weather Condition")
ax.set_ylabel("Avg Delivery Time (min)")
plt.tight_layout()
plt.savefig("../output/charts/08_delivery_by_weather.png", dpi=150)
plt.show()
```
